# Panorama dos Modelos Supervisionados

Os algoritmos de aprendizado supervisionado dividem-se em dois grupos pelo tipo de saída:
- **Regressão** → saída contínua
- **Classificação** → saída categórica

Alguns modelos operam em ambos os regimes (SVM, Árvore de Decisão, Random Forest etc.).

| Modelo | Tipo | Regressão | Classificação | Quando usar |
|---|---|---|---|---|
| Regressão Linear | Regressão | Prever valor contínuo (ex: preço de imóvel) | — | Baseline rápido; relação linear; interpretabilidade |
| Regressão Logística | Classificação | — | Saída probabilística via sigmoide; binário ou OvR | Classificação binária interpretável; probabilidade da classe importa |
| k-NN | Ambos | Média ponderada dos k vizinhos | Votação dos k vizinhos | Datasets pequenos; fronteira irregular; não-paramétrico |
| SVM | Ambos | SVR: hiperplano dentro de ε-tube; robusto a outliers | SVC: maximiza margem; kernel trick para não-lineares | Alta dimensão; datasets pequenos/médios; fronteiras complexas |
| Árvore de Decisão | Ambos | Partições recursivas; minimiza MSE na folha | Critério Gini ou Entropia | Dados mistos; interpretabilidade; sem necessidade de normalização |
| Random Forest | Ambos | Média de múltiplas árvores (bagging) | Votação majoritária; robusto a overfitting | Datasets médios/grandes; feature importance gratuita |
| Gradient Boosting | Ambos | Aprende resíduo a cada iteração | Ensemble de árvores fracas | Dados tabulares; acurácia máxima |
| Naïve Bayes | Classificação | — | Teorema de Bayes com independência condicional | NLP; datasets pequenos; alta dimensionalidade esparsa |
| Redes Neurais (MLP) | Ambos | Aproximador universal; saída linear | Softmax na última camada | Imagem, texto, séries temporais; grandes volumes |
